## Data Quality Analysis

Before performing exploratory analysis, the dataset is checked for
missing values, duplicate records, and invalid transaction values.

This step helps identify data quality issues that may affect the
accuracy of the analysis.

```python
missing_values=df.isna().sum()
missing_values
```

## Removing Duplicate Records

Duplicate rows represent repeated transaction records. These records
can lead to double counting during analysis, so they are removed.

```python
duplicate_count = df.duplicated().sum()
duplicate_count
```

```python
print("Number of duplicate rows:" , duplicate_count)
```

```python
df=df.drop_duplicates()
df
```

## Standardizing Column Names

The original dataset contains some column names that are either generic
or contain spaces. These names are standardized to make the dataset
clearer, more descriptive, and easier to work with during analysis.

For example, `Price` represents the unit price of a product, while
`Customer ID` is renamed to a consistent format without spaces.

```python
df=df.rename(columns={"Invoice":"InvoiceNo","Price":"UnitPrice","Customer ID":"CustomerID"})
df
```

## Handling Missing Product Descriptions

Some transaction records have missing product descriptions. Since the
product description is useful for product-level analysis, these missing
values are examined before deciding how to handle them.

The number of missing descriptions is checked to understand their impact
on the dataset.

```python
df["Description"].isna().sum()
```

## Removing Records with Missing Descriptions

Product descriptions are required for product-level analysis.
Since only a small number of records have missing descriptions,
these records are removed from the dataset.

This prevents incomplete product information from affecting
the analysis.

```python
df[df["Description"].isna()]
```

```python
df[df["Description"].isna()][["StockCode", "Description"]].head(10)
```

```python
description_by_stockcode=df.groupby("StockCode")["Description"].first()
description_by_stockcode
```

```python
df["Description"] = df["StockCode"].map(description_by_stockcode)
```

```python
df["Description"].isna().sum()
```

## Removing Unrecoverable Product Descriptions

Some product descriptions could not be recovered using the corresponding
StockCode because no valid description was available elsewhere in the
dataset.

These remaining records are removed because a missing product description
would limit product-level analysis.

```python
df=df.dropna(subset='Description')
df
```

```python
df["Description"].isna().sum()
```

## Handling Invalid Quantity Values

The Quantity column represents the number of units involved in a transaction.
Negative quantities can occur in this dataset because of cancelled or
returned transactions.

These values should be examined before deciding how they should be handled,
so that valid cancellation information is not accidentally removed.

```python
df.Quantity.min()
```

```python
(df["Quantity"] <=0).sum()
```

```python
df["InvoiceNo"].str.startswith("C").sum()
```

## Handling Invalid Unit Prices

UnitPrice represents the price of one unit of a product.
Zero or negative prices are not valid for revenue calculations,
so these values need to be identified before calculating revenue.

Cancellation records are retained because they may be useful for
separate cancellation analysis.

```python
df.UnitPrice.min()
```

```python
(df.UnitPrice <=0 ).sum()
```

```python
df = df[df["UnitPrice"]>0]
df
```

```python
df["UnitPrice"].min()
```

## Creating Revenue Feature

Revenue represents the total value of each transaction line.
It is calculated by multiplying the quantity of products sold by
the unit price.

This feature will be used later to analyze total sales, product
performance, country-wise revenue, and revenue trends over time.

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

```python
df.head()
```

## Creating Date and Time Features

The InvoiceDate column contains the date and time of each transaction.
Additional date and time features are created to make it easier to
analyze sales patterns across years, months, days, and hours.

These features will later help identify seasonal and time-based
purchasing patterns.

```python
df["Year"] = df["InvoiceDate"].dt.year
```

```python
df[["InvoiceDate" , "Year"]].head()
```

```python
df["Month"] = df["InvoiceDate"].dt.month
```

```python
df[["InvoiceDate" , "Month"]].head()
```

```python
df["MonthName"] = df["InvoiceDate"].dt.month_name()
```

```python
df[["InvoiceDate", "MonthName"]].head()
```

```python
df["Day"] = df["InvoiceDate"].dt.day
```

```python
df[["InvoiceDate", "Day"]].head()
```

```python
df["DayName"] = df["InvoiceDate"].dt.day_name()
```

```python
df[["InvoiceDate", "DayName"]].head()
```

```python
df["Hour"] = df["InvoiceDate"].dt.hour
```

```python
df[["InvoiceDate", "Hour"]].head()
```

## Creating Cancellation Flag

Some transactions in the dataset represent cancellations.
These transactions can be identified by InvoiceNo values that start
with the letter `C`.

A binary cancellation flag is created to make cancellation analysis
easier during exploratory analysis and dashboard development.

```python
df["IsCancelled"] = df["InvoiceNo"].str.startswith("C")
```

```python
df["IsCancelled"].value_counts()
```
